```python
# Problem Statement
# Create a simple interactive app inside a Jupyter Notebook where a user types their name
# and the app returns a personalized greeting.

# Step 2: Import the Gradio library
import gradio as gr

# Step 3: Define core business logic (Python function)
def greet(name):
    """
    Takes a person's name as input and returns a greeting message.

    Parameters:
    name (str): Name entered by the user

    Returns:
    str: A friendly greeting
    """
    return f"Hello {name}, welcome to your first Gradio app inside Jupyter Notebook!"

# Step 4: Create Gradio Interface
demo = gr.Interface(
    fn=greet,                           # link our function
    inputs=gr.Textbox(label="Enter Your Name"),
    outputs=gr.Textbox(label="Greeting"),
    title="Lesson 1: Greeting App",
    description="Type your name and get a personalized greeting."
)

# Step 5: Launch the app in notebook mode
demo.launch(share=True)
# Important:
# share=False → keeps it local
# If you want a public URL: share=True
```

```python
# Problem Statement

# Build an app inside Jupyter Notebook that:
# Accepts three different inputs:
# Name (Textbox)
# Age Group (Dropdown)
# Is the user a student? (Checkbox)
# Returns a custom message based on all inputs.

# Goal: Learn how Gradio handles lists of inputs, how data flows into your Python function, and how to build slightly more realistic apps.

# Step 1: Import Gradio
import gradio as gr

# Step 2: Define the core business logic

def user_profile(name, age_group, is_student):
    """
    Builds a personalized message using multiple inputs.

    Parameters:
    name (str): The user's name
    age_group (str): The selected age group
    is_student (bool): Whether the user is a student or not

    Returns:
    str: Personalized message describing the user
    """

    student_status = "a student" if is_student else "not a student"

    return f"Hello {name}! You belong to the {age_group} age group and you are {student_status}."

# Step 3: Build the Gradio Interface for multiple inputs

demo = gr.Interface(
    fn=user_profile,
    inputs=[
        gr.Textbox(label="Enter Your Name"),                   # Input 1
        gr.Dropdown(choices=["18-25", "26-35", "36-50", "50+"],
                    label="Select Your Age Group"),            # Input 2
        gr.Checkbox(label="Are you a student?")                # Input 3
    ],
    outputs=gr.Textbox(label="Personalized Output"),
    title="Lesson 2: Multiple Inputs Demo",
    description="This app demonstrates how to use multiple input components in Gradio."
)

# Step 4: Launch inside Jupyter Notebook
demo.launch()
```

```python
# Problem Statement

# Build a Gradio app that takes a sentence from the user and returns:
# The sentence in uppercase (text output)
# The word count (number output)
# A word-cloud-like image generated dynamically (image output)

# This will teach:
# How to return multiple values
# How to use multiple output components
# How to handle images generated inside Python

# Step 1: Import required libraries
import gradio as gr
from PIL import Image, ImageDraw, ImageFont
import numpy as np

# Step 2: Define the core function with multiple outputs

def process_text(sentence):
    """
    Takes a sentence and returns:
    1. Uppercase sentence
    2. Word count
    3. Generated image containing the text
    """

    # --- 1. Convert to uppercase ---
    upper_text = sentence.upper()

    # --- 2. Count words ---
    word_count = len(sentence.split())

    # --- 3. Generate a simple image with the text ---
    img = Image.new("RGB", (600, 200), color="white")  # blank white image
    draw = ImageDraw.Draw(img)

    # Default PIL fonts — avoids font loading issues in notebooks
    draw.text((10, 80), upper_text, fill="black")

    # Convert PIL image to a NumPy array (Gradio accepts np.array)
    img_array = np.array(img)

    # Return the 3 outputs
    return upper_text, word_count, img_array

# Step 3: Create Gradio Interface with multiple outputs

demo = gr.Interface(
    fn=process_text,  # our function
    inputs=gr.Textbox(label="Enter a Sentence"),
    outputs=[
        gr.Textbox(label="Uppercase Text"),   # Output 1
        gr.Number(label="Word Count"),        # Output 2
        gr.Image(label="Generated Image")     # Output 3
    ],
    title="Lesson 3: Multiple Outputs Demo",
    description="This app demonstrates how to return multiple outputs using Gradio."
)

# Step 4: Launch inside Jupyter Notebook
demo.launch()
```

```python
# Problem Statement

# Build a mini app that uses Gradio Blocks to create a layout where:
# Left side → input textbox + button
# Right side → output text
# When button is clicked → text is reversed

# This will teach you:
# gr.Blocks()
# gr.Row()
# gr.Column()
# Component events → .click()
# How to build flexible layouts

# Step 1: Import Gradio
import gradio as gr

# Step 2: Define the function that will be triggered by the button
def reverse_text(text):
    """
    Takes text input and returns the reversed text.
    """
    return text[::-1]

# Step 3: Build the layout using Blocks

with gr.Blocks(title="Lesson 4: Blocks Basics") as demo:

    # Create a row (left side inputs, right side outputs)
    with gr.Row():

        # Left side column (inputs)
        with gr.Column():
            gr.Markdown("### Enter Text to Reverse")

            input_box = gr.Textbox(label="Your Text")
            reverse_btn = gr.Button("Reverse Text")

        # Right side column (outputs)
        with gr.Column():
            gr.Markdown("### Reversed Output")

            output_box = gr.Textbox(label="Reversed Text")

    # Define what happens when the button is clicked
    reverse_btn.click(
        fn=reverse_text,     # the function to call
        inputs=input_box,    # input component
        outputs=output_box   # output component
    )


# Step 4: Launch the Blocks app
demo.launch()
```

```python
# -------------------- PDF → TEXT --------------------

def load_pdf_chunks(pdf_file, chunk_size=300): # Adjust chunk size as needed
    if pdf_file is None: # Handle case where no file is uploaded
        return [] # Return empty list if no file is uploaded
    # pdf_stream = BytesIO(pdf_file.read())
    # reader = PyPDF2.PdfReader(pdf_stream)
    reader = PyPDF2.PdfReader(pdf_file.name) # Corrected to read from file path
    text = "" # Extract text from all pages
    for page in reader.pages: # Iterate through each page
        page_text = page.extract_text() # Extract text from the page
        if page_text: # Check if text is not None
            text += page_text.replace("\n", " ") + " " # Replace newlines with spaces
    chunks = [text[i:i + chunk_size] for i in range(0, len(text), chunk_size)]
    # if chunksize = 500, the:
    # Iteration1: text[0:500]
    # Iteration2: text[500:1000]
    # and so on...
    return chunks # Return list of text chunks

# -------------------- SETUP RAG --------------------

def setup_rag(chunks): # Setup embedding model, FAISS index, and QA pipeline
    if not chunks: # Handle case where chunks list is empty
        return None, None, None, None # Return None if no chunks are available
    embedder = SentenceTransformer("all-MiniLM-L6-v2") # Embedding model
    # What is all-MiniLM-L6-v2?
    # It's a compact, efficient model for generating sentence embeddings.
    # The number of embedding dimensions is 384.
    # It is good for semantic search and clustering tasks.
    # Ofcourse, it is not as powerful as larger models, but it's fast and lightweight.
    chunk_embeddings = embedder.encode(chunks, convert_to_tensor=False)
    dim = len(chunk_embeddings[0]) # Dimension of embeddings i.e. 384
    # Initialize FAISS index
    index = faiss.IndexFlatL2(dim) # L2 distance metric
    index.add(np.array(chunk_embeddings).astype("float32")) # Add embeddings to index
    # Setup QA pipeline
    qa_pipeline = pipeline("text2text-generation", model="google/flan-t5-base", max_length=256) # Text generation model
    # google/flan-t5-base is a fine-tuned version of T5 for instruction-based tasks.
    # It is good for generating answers based on provided context.
    return chunks, embedder, index, qa_pipeline # Return all components for RAG

# -------------------- RETRIEVAL --------------------

def retrieve_top_k(query, chunks, embedder, index, k=3): # Retrieve top-k relevant chunks for the query
    query_embedding = embedder.encode([query]) # Embed the query
    distances, indices = index.search(np.array(query_embedding).astype("float32"), k) # Search FAISS index
    return [chunks[i] for i in indices[0]] # Return top-k chunks

# -------------------- ANSWERING --------------------
def answer_query(query, chunks, embedder, index, qa_pipeline): # Answer the query using retrieved chunks and QA pipeline
    top_chunks = retrieve_top_k(query, chunks, embedder, index) # Get top-k relevant chunks
    context = "\n".join(top_chunks) # Combine chunks into context
    prompt = f"Context: {context}\n\nQuestion: {query}\n\nAnswer:" # Create prompt for QA model
    response = qa_pipeline(prompt)[0]["generated_text"] # Generate answer
    return response.strip() # Return the answer

# -------------------- GRADIO UI --------------------

rag_state = {
    "chunks": None,
    "embedder": None,
    "index": None,
    "qa_pipeline": None,
    "chat_history": []
}


def handle_file_upload(file):
    if file is None: # Handle case where no file is uploaded
        return "❌ No file uploaded." # Return error message if no file is uploaded

    file_name = os.path.basename(file.name) # Get the file name
    file_size = os.path.getsize(file.name) # Get the file size
    return f"✅ File uploaded: {file_name}\n📦 Size: {file_size} bytes. \tClick on \"Initialize RAG\" button"

def initialize_rag(pdf_file): # Initialize RAG system with uploaded PDF
    if pdf_file is None: # Handle case where no file is uploaded
        return "❌ No PDF uploaded." # Return error message if no file is uploaded

    chunks = load_pdf_chunks(pdf_file) # Load and chunk PDF text
    chunks, embedder, index, qa_pipeline = setup_rag(chunks) # Setup RAG components
    rag_state.update({
        "chunks": chunks,
        "embedder": embedder,
        "index": index,
        "qa_pipeline": qa_pipeline,
        "chat_history": []
    }) # Update global RAG state
    return "✅ PDF loaded and RAG system initialized." # Return success message

def chat(query): # Handle user query and generate response
    if not query.strip(): # Check for empty query
        return "⚠️ Please enter a valid question." # Prompt user to enter a valid question
    if rag_state["chunks"] is None: # Check if RAG is initialized
        return "⚠️ Please upload and initialize a PDF first." # Prompt user to upload and initialize PDF
    response = answer_query(query, rag_state["chunks"], rag_state["embedder"], rag_state["index"], rag_state["qa_pipeline"]) # Get answer from RAG system
    rag_state["chat_history"].append(("🧑 You", query)) # Append user query to chat history
    rag_state["chat_history"].append(("🤖 Assistant", response)) # Append assistant response to chat history
    return "\n\n".join([f"**{speaker}**: {text}" for speaker, text in rag_state["chat_history"]]) # Format chat history for display

with gr.Blocks() as demo:
    gr.Markdown("# 🤖 HR FAQ Assistant (Naive RAG)")
    gr.Markdown("Upload a company policy PDF and ask HR-related questions.")

    with gr.Row():
        pdf_input = gr.File(label="Upload PDF", file_types=[".pdf"])
        init_button = gr.Button("Initialize RAG")

    status = gr.Textbox(lines=2,label="Status -", interactive=False)
    pdf_input.change(fn=handle_file_upload, inputs=pdf_input, outputs=status)

    query_input = gr.Textbox(lines=2, label="Question-", placeholder="Ask your HR-related question...")
    submit_btn = gr.Button("Submit")  # add submit button
    response_output = gr.Markdown()

    init_button.click(fn=initialize_rag, inputs=pdf_input, outputs=status)
    query_input.submit(fn=chat, inputs=query_input, outputs=response_output)
    submit_btn.click(fn=chat, inputs=query_input, outputs=response_output ) # Handler for Submit button

demo.launch(share=True)
```

```python
import gradio as gr
from openai import OpenAI
import os

# -----------------------------
# OpenAI Client Setup
# -----------------------------
from google.colab import userdata
OPENAI_API_KEY = userdata.get('OPENAI_API_KEY')
client = OpenAI(
    api_key=OPENAI_API_KEY
)

# -----------------------------
# Core Image Generation Logic
# -----------------------------
def generate_image(prompt):
    """
    Takes a text prompt and generates an image using OpenAI DALL·E.
    Returns an image URL that Gradio can display.
    """

    if not prompt or prompt.strip() == "":
        return None, "❌ Please enter a valid prompt."

    response = client.images.generate(
        model="dall-e-2",
        prompt=prompt,
        size="1024x1024",
        n=1
    )

    image_url = response.data[0].url
    return image_url, "✅ Image generated successfully."


# -----------------------------
# Gradio UI
# -----------------------------
with gr.Blocks(title="Text to Image Generator (OpenAI)") as demo:

    gr.Markdown("# 🎨 Text → Image Generator")
    gr.Markdown("Enter a prompt and generate an image using **OpenAI DALL·E**.")

    with gr.Row():
        with gr.Column():
            prompt_input = gr.Textbox(
                label="Enter Image Prompt",
                placeholder="A white Persian cat sitting on a crimson pillow...",
                lines=4
            )
            generate_btn = gr.Button("Generate Image")

        with gr.Column():
            image_output = gr.Image(
                label="Generated Image",
                height=512
            )
            status_output = gr.Markdown()

    generate_btn.click(
        fn=generate_image,
        inputs=prompt_input,
        outputs=[image_output, status_output]
    )

demo.launch()
```
