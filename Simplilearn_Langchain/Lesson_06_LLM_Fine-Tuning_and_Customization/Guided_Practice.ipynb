### **Step 2: Generate image from text**
- Define a text prompt describing the image you want to generate
- Use the OpenAI Images API to generate an image using the DALL·E model — "gpt-image-1"
- Decode and save the image locally

```python
from openai import OpenAI
import base64
import os

# Initialize client (ensure your API key is set as an environment variable)
client = OpenAI()

# Prompt user for input
text_prompt = input("Enter a description of a historical landmark: ")

try:
    # Generate image (✅ Correct size format)
    response = client.images.generate(
        model="gpt-image-1",
        prompt=text_prompt,
        size="1024x1024"
    )

    # Check if image data exists
    if response.data and response.data[0].b64_json:
        image_base64 = response.data[0].b64_json
        image_bytes = base64.b64decode(image_base64)

        # Save image locally
        with open("generated_image.png", "wb") as f:
            f.write(image_bytes)

        print("✅ Image successfully generated and saved as 'generated_image.png'")
    else:
        print("⚠️ No image data returned. Check your prompt or API key.")

except Exception as e:
    print(f"❌ Error generating image: {e}")
```

### **Step 3: Display the generated images**

- Load the generated image using Pillow (PIL)
- Display it inline using Matplotlib for visual confirmation

```python
from openai import OpenAI
import base64
from IPython.display import Image, display

# Initialize OpenAI client (make sure your API key is set in environment)
client = OpenAI()

# Define landmark prompts
landmark_prompts = [
    "The Pyramids of Giza at sunset",
    "The Colosseum in Rome on a sunny day",
    "The Taj Mahal reflecting in water at dawn",
    "Machu Picchu surrounded by clouds"
]

# Generate and display images for each prompt
for i, prompt in enumerate(landmark_prompts, start=1):
    print(f"\n🎨 Generating image {i} for: '{prompt}'")

    try:
        # ✅ Correct size format: "widthxheight"
        response = client.images.generate(
            model="gpt-image-1",
            prompt=prompt,
            size="1024x1024"
        )

        # Validate and decode response
        if response.data and response.data[0].b64_json:
            image_base64 = response.data[0].b64_json
            image_bytes = base64.b64decode(image_base64)

            # Save locally
            image_path = f"landmark_{i}.png"
            with open(image_path, "wb") as f:
                f.write(image_bytes)

            print(f"✅ Image saved as '{image_path}'")

            # Display inline
            display(Image(image_bytes))
        else:
            print("⚠️ No image data returned. Check your API key or prompt.")
    except Exception as e:
        print(f"❌ Error generating image: {e}")
```
