### __Step 1: Define a template using the f-string format__
- Define a string template using Python’s **f-string** formatting approach  
- Use placeholders like `{book_title}` to dynamically insert variable values into the prompt at runtime
  ```python


````python
fstring_template = "Here is a brief summary for the book titled '{book_title}':"
````

This template allows for flexible substitution, making it easy to reuse for multiple books or content items.

### __Step 2: Create a prompt using the f-string template__
- Instantiate the template with a specific book title to generate a customized prompt for summarization.

````python
book_title = "The Great Gatsby"

prompt = fstring_template.format(book_title=book_title)
````

- In this step, the placeholder {book_title} is replaced dynamically with the given title to form a complete and context-specific prompt.

### __Step 3: Use the prompt to generate a question__
- Define a function to simulate or send the prompt for summary generation



````python
def get_book_summary(prompt):

    return "It's a novel about love, wealth, and aspiration, set in the Roaring '20s."

summary = get_book_summary(prompt)
print(summary)
````

- This step demonstrates how a function can use the dynamically created prompt to fetch or generate relevant text based on the template.

### __Step 4: Output the summary__
- Display the generated summary to the user or save it to a file or database, depending on the application

````python
print(f"Summary for '{book_title}': {summary}")
````

````python
from string import Template
template = Template("Hello, my name is $name and I am $age years old")
formatted_template = template.substitute(name = "ABC", age = 30)
formatted_template
````
