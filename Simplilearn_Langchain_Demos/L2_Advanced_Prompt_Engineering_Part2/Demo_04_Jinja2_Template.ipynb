### __Step 1: Set up the OpenAI API key__
- Import the required libraries and set up the OpenAI API key
- The os library interacts with the system, openai connects to the OpenAI API, and PromptTemplate from LangChain enables template-based prompt creation.

```python
import os
from openai import OpenAI
from langchain_core.prompts import PromptTemplate

client = OpenAI()
```

### __Step 2: Define a function to get completion__
- Define a helper function to handle the interaction with the OpenAI API
- Call the __client.chat.completions.create__ method to get a response from the model
- The temperature parameter is set to __0__ for deterministic (non-random) responses

```python
def get_completion(prompt, model="gpt-3.5-turbo"):
    messages = [{"role": "user", "content": prompt}]
    response = client.chat.completions.create(
        model=model,
        messages=messages,
        temperature=0
    )
    return response.choices[0].message.content
```

### __Step 3: Define your prompts__
- Create a simple string prompt template using LangChain’s PromptTemplate
- This approach demonstrates how LangChain can dynamically substitute input variables into templates to generate structured and repeatable prompts.

```python
# String Prompt Template
from langchain_core.prompts import PromptTemplate

templatee = " Please write a {length} review, of the book {book_title}. "

input_variabless = ["length", "book_title"]

# Creating the prompt template
prompt = PromptTemplate(
    input_variables=input_variabless,
    template=templatee
)

# Format the prompt
formatted_prompt = prompt.format(
    length="short",
    book_title="House Of Dragon"
)

print(formatted_prompt)

response = get_completion(formatted_prompt)
print("AI Response:")
print(response)
```

### __Step 4: Define Jinja2 prompt template__
- Create a Jinja2-style template using LangChain and pass user-defined variables dynamically
- Format it with user-provided variables
- Pass it to the AI model

```python


jinja2_template = "Give me an {{ adjective }} fact about {{ topic }}"

prompt = PromptTemplate.from_template(
    jinja2_template,
    template_format="jinja2"
)

user_question = prompt.format(
    adjective="interesting",
    topic="space exploration"
)

print(user_question)

response = get_completion(user_question)
print("AI Response:")
print(response)
```

- You can also use conditional logic in Jinja2 templates for adaptive prompt generation

```python
jinja2_template = """

Dear {{ name }},
{% if age < 18 %}
You are invited to our kids' event with activities such as face painting, bouncy castles, and clown shows.
{% elif age < 65 %}
You are invited to our adult event with activities like live music, wine tasting, and art workshops.
{% else %}
You are invited to our senior event with activities including book clubs, chess tournaments, and tea dances.
{% endif %}
Sincerely,
Event Organizer

Write the mail in 200 words
"""

prompt = PromptTemplate.from_template(
    jinja2_template,
    template_format="jinja2"
)

argument_prompt = prompt.format(
    name="John Doe",
    age=12
)

print(argument_prompt)

```

```python
response = get_completion(argument_prompt)
print("AI Response:")
print(response)
```

- This example shows how Jinja2 enables flexible, logic-based prompt construction that adjusts content dynamically based on input values.
