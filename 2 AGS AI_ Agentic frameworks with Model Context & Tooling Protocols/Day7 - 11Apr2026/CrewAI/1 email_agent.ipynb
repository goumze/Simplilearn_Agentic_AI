### Problem Statement

In many organizations, employees often write emails in an informal, unclear, or unstructured manner, especially during quick communication. These emails may contain abbreviations, incomplete thoughts, or lack professionalism, which can lead to misunderstandings, poor communication, and reduced business impact.

The objective of this task is to build an AI-powered Email Assistant Agent that can automatically transform such rough, informal emails into well-structured, professional, and clear communication.

The system should:
- Take a raw, unpolished email as input
- Understand the intent and context of the message
- Expand abbreviations and incomplete phrases
- Improve tone, clarity, and professionalism
- Output a polished, business-ready email with proper formatting

This solution leverages an LLM-based agent using the CrewAI framework to simulate a communication expert who refines emails for professional use.

```python
from dotenv import load_dotenv
load_dotenv()
```

```python
# !pip install crewai -q
```

```python
from crewai import LLM

llm = LLM(
    model="gpt-4o",
    temperature=0.7,
    max_tokens=150,
)
```

```python
from crewai import Agent, Task, Crew
# Agent - An agent is an autonomous entity that can perform tasks, make decisions, and interact with its environment. It can be designed to achieve specific goals and can operate independently or in collaboration with other agents.
# Task - A task is a specific piece of work or an assignment that an agent is designed to accomplish. Tasks can vary in complexity and can be simple, such as answering a question, or more complex, such as managing a project or conducting research.
# Crew - A crew is a group of agents that work together to achieve a common goal. Crews can be organized hierarchically, with different agents taking on specific roles and responsibilities within the group. Crews can collaborate, share information, and coordinate their actions to accomplish tasks more efficiently.

email_assistant = Agent(
    role="Email Assistant Agent",
    goal="Improve emails and make them sound professional and clear",
    backstory="A highly experienced communication expert skilled in professional email writing",
    verbose=True, # Set to True to enable detailed logging of the agent's thought process and actions
    llm=llm
)
```

```python
original_email = """
hey team, just wanted to tell u that the demo is kind of ready, but there's still stuff left.
Maybe we can show what we have and say rest is WIP.
Let me know what u think. thanks
"""

email_task = Task(
    description=f"""Take the following rough email and rewrite it into a professional and polished version.
    Expand abbreviations:
    '''{original_email}'''""",
    agent=email_assistant,
    expected_output="A professional written email with proper formatting and content.",
)
```

```python
crew = Crew(
    agents=[email_assistant], # eg: agents=[email_assistant, grammar_checker, tone_fixer]
    tasks=[email_task],
    verbose=True
)
result = crew.kickoff()
# kickoff() - This method starts the execution of the crew's tasks. It initiates the agents to perform their assigned tasks and coordinates their interactions to achieve the desired outcomes. The kickoff() method typically triggers the agents to begin processing their tasks, generating outputs, and collaborating as needed to accomplish the goals set for the crew.
print(result)
```
