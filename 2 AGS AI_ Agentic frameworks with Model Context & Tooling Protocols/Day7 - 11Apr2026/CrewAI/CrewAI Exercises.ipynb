```python
# Problem Statement: RFP (Request for Proposal) Auto-Response System

# Organizations receive detailed RFP documents from clients that require structured, accurate, and tailored responses within tight deadlines.

# The challenge is to:
# - Understand and extract key requirements from RFP documents
# - Map requirements to company capabilities
# - Generate a well-structured proposal response
# - Ensure clarity, completeness, and professionalism

# We will build a CrewAI-based multi-agent system where:
# - One agent extracts and structures requirements from the RFP
# - One agent maps requirements to solutions and capabilities
# - One agent generates a proposal response
# - A manager agent reviews and finalizes the output

# This system automates the proposal creation workflow, reducing manual effort and improving response quality.
```

```python
# ============================================================
# USE CASE: RFP Auto-Response System (CrewAI - Enterprise)
# ============================================================

# Step 1: Install dependencies
# !pip install crewai -q

# Step 2: Imports
from dotenv import load_dotenv
load_dotenv()

import os
from crewai import Agent, Task, Crew, Process, LLM

# Step 3: LLM Config
llm = LLM(
    model="gpt-4o",
    temperature=0.4
)

# ============================================================
# Step 4: Define Agents
# ============================================================

# Agent 1: Requirement Extractor
extractor_agent = Agent(
    role="RFP Requirement Analyst",
    goal="Extract and structure key requirements from RFP document",
    backstory="Expert in analyzing business and technical RFP documents",
    llm=llm,
    verbose=True
)

# Agent 2: Solution Mapper
solution_agent = Agent(
    role="Solution Architect",
    goal="Map extracted requirements to company capabilities and solutions",
    backstory="Experienced solution architect with deep knowledge of enterprise systems",
    llm=llm,
    verbose=True
)

# Agent 3: Proposal Writer
writer_agent = Agent(
    role="Proposal Writer",
    goal="Create a professional and structured proposal response",
    backstory="Expert in writing winning business proposals",
    llm=llm,
    verbose=True
)

# Agent 4: Manager (Supervisor)
manager_agent = Agent(
    role="Proposal Manager",
    goal="Ensure proposal quality, completeness, and alignment with requirements",
    backstory="Senior manager responsible for final proposal approval",
    llm=llm,
    allow_delegation=True,
    verbose=True
)

# ============================================================
# Step 5: Define Tasks
# ============================================================

rfp_document = """
Client requires an AI-based demand forecasting system.
- Must integrate with existing ERP
- Should provide real-time insights
- Needs dashboard reporting
- Timeline: 8 weeks
"""

# Task 1: Extract requirements
task1 = Task(
    description=f"Extract structured requirements from the following RFP:\n{rfp_document}",
    expected_output="Structured list of requirements with categories",
    agent=extractor_agent
)

# Task 2: Map solution
task2 = Task(
    description="Map the extracted requirements to a suitable technical solution and approach",
    expected_output="Detailed solution mapping aligned to requirements",
    agent=solution_agent,
    context=[task1]
)

# Task 3: Write proposal
task3 = Task(
    description="Write a professional proposal including solution, approach, deliverables, and timeline",
    expected_output="Well-structured proposal document",
    agent=writer_agent,
    context=[task2]
)

# ============================================================
# Step 6: Crew Setup (Hierarchical)
# ============================================================

crew = Crew(
    agents=[extractor_agent, solution_agent, writer_agent],
    tasks=[task1, task2, task3],
    process=Process.hierarchical,
    manager_agent=manager_agent,
    verbose=True
)

# ============================================================
# Step 7: Execute
# ============================================================

result = crew.kickoff()

print("\n================ FINAL PROPOSAL OUTPUT ================\n")
print(result)
```

```python
# Problem Statement: Incident RCA (Root Cause Analysis) Generator

# Organizations frequently face production incidents such as system failures, performance degradation, and service outages. After resolution, a detailed Root Cause Analysis (RCA) report must be created.

# The challenge is to:
# - Analyze logs and incident details
# - Identify the root cause of the issue
# - Suggest corrective and preventive actions
# - Generate a structured RCA report

# We will build a CrewAI-based multi-agent system where:
# - One agent analyzes logs and extracts key issues
# - One agent identifies root cause
# - One agent suggests corrective actions
# - A manager agent validates and produces a final RCA report

# This system automates incident analysis and reporting, reducing manual effort and improving consistency.
```

```python
# ============================================================
# USE CASE: Incident RCA Generator (CrewAI - Enterprise)
# ============================================================

# Step 1: Install dependencies
# !pip install crewai -q

# Step 2: Imports
from dotenv import load_dotenv
load_dotenv()

from crewai import Agent, Task, Crew, Process, LLM

# Step 3: LLM Config
llm = LLM(
    model="gpt-4o",
    temperature=0.3
)

# ============================================================
# Step 4: Define Agents
# ============================================================

# Agent 1: Log Analyzer
log_agent = Agent(
    role="Log Analysis Specialist",
    goal="Analyze incident logs and extract key error patterns",
    backstory="Expert in analyzing system logs and identifying anomalies",
    llm=llm,
    verbose=True
)

# Agent 2: Root Cause Analyst
rca_agent = Agent(
    role="Root Cause Analyst",
    goal="Identify root cause of the incident based on log analysis",
    backstory="Experienced SRE with expertise in incident debugging",
    llm=llm,
    verbose=True
)

# Agent 3: Fix & Prevention Agent
fix_agent = Agent(
    role="Remediation Specialist",
    goal="Suggest corrective and preventive actions",
    backstory="Expert in system reliability and preventive measures",
    llm=llm,
    verbose=True
)

# Agent 4: Manager (Supervisor)
manager_agent = Agent(
    role="Incident Manager",
    goal="Validate RCA report and ensure completeness and clarity",
    backstory="Senior incident manager responsible for final RCA reporting",
    llm=llm,
    allow_delegation=True,
    verbose=True
)

# ============================================================
# Step 5: Define Tasks
# ============================================================

incident_data = """
Incident Details:
- API response time increased to 15 seconds
- Database timeout errors observed
- High traffic during sale event
- Some requests failing intermittently
"""

# Task 1: Analyze logs
task1 = Task(
    description=f"Analyze the following incident logs and extract key issues:\n{incident_data}",
    expected_output="Structured summary of observed issues",
    agent=log_agent
)

# Task 2: Identify root cause
task2 = Task(
    description="Identify root cause based on log analysis",
    expected_output="Clear root cause explanation",
    agent=rca_agent,
    context=[task1]
)

# Task 3: Suggest fixes
task3 = Task(
    description="Suggest corrective and preventive actions",
    expected_output="Actionable fix and prevention steps",
    agent=fix_agent,
    context=[task2]
)

# ============================================================
# Step 6: Crew Setup (Hierarchical)
# ============================================================

crew = Crew(
    agents=[log_agent, rca_agent, fix_agent],
    tasks=[task1, task2, task3],
    process=Process.hierarchical,
    manager_agent=manager_agent,
    verbose=True
)

# ============================================================
# Step 7: Execute
# ============================================================

result = crew.kickoff()

print("\n================ FINAL RCA REPORT ================\n")
print(result)
```

```python
# Problem Statement: Regulatory Impact Analyzer

# Organizations regularly receive regulatory circulars, policy updates, and compliance documents. However, only some parts of these documents are relevant to the company’s operations.

# The challenge is to:
# - Analyze large regulatory documents
# - Identify relevant sections for the organization
# - Understand business impact
# - Suggest actionable next steps

# We will build a CrewAI-based multi-agent system where:
# - One agent extracts key points from regulatory documents
# - One agent filters relevance based on company context
# - One agent analyzes business impact
# - One agent suggests actions
# - A manager agent validates and generates a final impact report

# This system helps organizations quickly understand and act on regulatory changes.
```

```python
# ============================================================
# USE CASE: Regulatory Impact Analyzer (CrewAI - Enterprise)
# ============================================================

# Step 1: Install dependencies
# !pip install crewai -q

# Step 2: Imports
from dotenv import load_dotenv
load_dotenv()

from crewai import Agent, Task, Crew, Process, LLM

# Step 3: LLM Config
llm = LLM(
    model="gpt-4o",
    temperature=0.3
)

# ============================================================
# Step 4: Define Agents
# ============================================================

# Agent 1: Document Analyzer
doc_agent = Agent(
    role="Regulatory Document Analyst",
    goal="Extract key points from regulatory documents",
    backstory="Expert in analyzing legal and regulatory documents",
    llm=llm,
    verbose=True
)

# Agent 2: Relevance Filter
filter_agent = Agent(
    role="Relevance Specialist",
    goal="Identify sections relevant to the company's operations",
    backstory="Understands business context and filters relevant information",
    llm=llm,
    verbose=True
)

# Agent 3: Impact Analyzer
impact_agent = Agent(
    role="Business Impact Analyst",
    goal="Analyze the impact of regulations on the organization",
    backstory="Expert in translating regulations into business impact",
    llm=llm,
    verbose=True
)

# Agent 4: Action Advisor
action_agent = Agent(
    role="Compliance Advisor",
    goal="Suggest actionable steps based on regulatory impact",
    backstory="Expert in compliance strategy and execution",
    llm=llm,
    verbose=True
)

# Agent 5: Manager
manager_agent = Agent(
    role="Compliance Manager",
    goal="Ensure report completeness and accuracy",
    backstory="Senior compliance officer responsible for final review",
    llm=llm,
    allow_delegation=True,
    verbose=True
)

# ============================================================
# Step 5: Input (Simulated Circular)
# ============================================================

regulatory_doc = """
New circular:
- Updated drug pricing guidelines introduced
- Stricter approval process for generic drugs
- Mandatory reporting requirements increased
- Applies to pharmaceutical manufacturers only
"""

company_context = "Pharmaceutical company focused on drug manufacturing and approvals"

# ============================================================
# Step 6: Define Tasks
# ============================================================

# Task 1: Extract key points
task1 = Task(
    description=f"Extract key regulatory points from document:\n{regulatory_doc}",
    expected_output="Structured key points",
    agent=doc_agent
)

# Task 2: Filter relevance
task2 = Task(
    description=f"Filter relevant points for company context: {company_context}",
    expected_output="Relevant regulatory items",
    agent=filter_agent,
    context=[task1]
)

# Task 3: Analyze impact
task3 = Task(
    description="Analyze business impact of relevant regulations",
    expected_output="Detailed business impact analysis",
    agent=impact_agent,
    context=[task2]
)

# Task 4: Suggest actions
task4 = Task(
    description="Suggest actionable compliance steps",
    expected_output="List of recommended actions",
    agent=action_agent,
    context=[task3]
)

# ============================================================
# Step 7: Crew Setup (Hierarchical)
# ============================================================

crew = Crew(
    agents=[doc_agent, filter_agent, impact_agent, action_agent],
    tasks=[task1, task2, task3, task4],
    process=Process.hierarchical,
    manager_agent=manager_agent,
    verbose=True
)

# ============================================================
# Step 8: Execute
# ============================================================

result = crew.kickoff()

print("\n================ FINAL REGULATORY IMPACT REPORT ================\n")
print(result)
```
