```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from crewai import LLM

llm = LLM(
    model="gpt-4o",
    temperature=0.7,
    max_tokens=150,
)
```

```python
import os
from crewai import Agent, Task, Crew
from crewai_tools import SerperDevTool # SerperDevTool is a search engine tool that connects to the Serper API, which is a Google Search API alternative similar to SerpAPI
# It is a prebuilt tool that lets your CrewAI agents perform real-time Google search.
# Under the hood, it sends your query to Serper.dev, a fast and cost-effective Google Search API.

research_agent = Agent(
    role="Research Specialist",
    goal="Research interesting facts about the topic: {topic}",
    backstory="You are an expert at finding relevant and factual data.",
    tools=[SerperDevTool()] if os.getenv("SERPER_API_KEY") else [],
    # https://docs.crewai.com/en/tools/search-research/serperdevtool
    # https://serper.dev/api-keys
    verbose=True,
    llm=llm
)

writer_agent = Agent(
    role="Creative Writer",
    goal="Write a short blog summary using the research",
    backstory="You are skilled at writing engaging summaries based on provided content.",
    llm=llm,
    verbose=True,
)
```

```python
task1 = Task(
    description="Find 3-5 interesting and recent facts about {topic} as of year 2025.",
    expected_output="A bullet list of 3-5 facts",
    agent=research_agent,
)

task2 = Task(
    description="Write a 100-word blog post summary about {topic} using the facts from the research.",
    expected_output="A blog post summary",
    agent=writer_agent,
    context=[task1],
)
```

```python
crew = Crew(
    agents=[research_agent, writer_agent],
    tasks=[task1, task2],
    verbose=True,
)

crew.kickoff(inputs={"topic": "The future of electrical vehicles"})
```
