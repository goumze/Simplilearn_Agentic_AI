## **Problem Statement**

Content creation workflows in modern digital teams often require two distinct skill sets: the ability to research accurate and up-to-date information, and the ability to convert that information into engaging, audience-ready content. Performing both tasks manually is time-consuming, inconsistent, and prone to human oversight.

To address this gap, we need a coordinated multi-agent system that can autonomously:

1. Research factual, relevant, and recent information about any topic.
2. Transform the researched content into a polished, high-quality written piece.
3. Maintain role specialization, where each agent focuses on one capability.
4. Pass context from one agent to another, ensuring continuity and accuracy across tasks.

This CrewAI system automates the entire research-to-content pipeline.
A Research Specialist agent first gathers factual insights based on the user’s topic.
A Creative Writer agent then uses those research findings to generate a clear, engaging, and publication-ready summary or blog post.

The final outcome is a streamlined, scalable workflow that reduces manual effort and ensures both factual integrity and high-quality writing.

```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from crewai import LLM

llm = LLM(
    model="gpt-4o",
    temperature=0.7,
    max_tokens=150,
)
```

```python
from crewai import Agent, Task, Crew

research_agent = Agent(
    role="Research Specialist",
    goal="Research interesting facts about the topic: {topic}",
    backstory="You are an expert at finding relevant and factual data.",
    verbose=True,
    llm=llm
)

writer_agent = Agent(
    role="Creative Writer",
    goal="Write a short blog summary using the research",
    backstory="You are skilled at writing engaging summaries based on provided content.",
    llm=llm,
    verbose=True,
)
```

```python
task1 = Task(
    description="Find 3-5 interesting and recent facts about {topic}.",
    expected_output="A bullet list of 3-5 facts",
    agent=research_agent,
)

task2 = Task(
    description="Write a 100-word blog post summary about {topic} using the facts from the research.",
    expected_output="A blog post summary",
    agent=writer_agent,
    context=[task1],
)
```

```python
crew = Crew(
    agents=[research_agent, writer_agent],
    tasks=[task1, task2],
    verbose=True,
)

crew.kickoff(inputs={"topic": "The future of electrical vehicles"})
```
