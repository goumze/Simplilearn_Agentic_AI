```python
import os
import openai
from langchain.chat_models import ChatOpenAI

from langchain.chains import ConversationChain
# pre-built chain designed for multi-turn dialogue.
# It automatically handles user inputs and model responses.

from langchain.memory import ConversationBufferMemory, ConversationBufferWindowMemory, ConversationTokenBufferMemory, ConversationSummaryBufferMemory
# Types of Memory in LangChain
# LangChain offers different memory types to suit various use cases:
# a. ConversationBufferMemory
# It keeps the full chat history from the start — nothing is forgotten.
# Best when: You want full context (e.g., RAG chat or tutoring assistant).
# Downside: Can become costly and hit token limits as conversation grows.

# b. ConversationBufferWindowMemory
# This one keeps only the last few exchanges (a sliding window).
# You specify a number like k=3 meaning “remember only last 3 messages.
# Older history is dropped to save cost and context space.
# Best when: You want short-term memory (like a chatbot that only cares about the latest context).
# Downside: Can lose important information if the conversation relies on earlier messages.

# c. ConversationTokenBufferMemory
# Similar to window memory, but instead of counting messages,
# it tracks number of tokens (words, subwords) to stay within model limits.
# You can set a max_token_limit, and old messages are dropped once that limit is reached.
# Best when: You want memory optimized for token-efficient chats
# (e.g., when using models with limited context window like 4K or 8K tokens).
# Downside: Can be complex to manage token counts and may still lose important context if not configured properly.

# d. ConversationSummaryBufferMemory
# Instead of storing all messages, it creates a running summary of old messages.
# When history grows too long, the model summarizes older parts and keeps only the essence.
# Best when: You want long-term conversation retention but still stay under token limits.
# (e.g., AI therapist, project assistant, or customer support agent).
# Downside: Summarization may lose details and nuance, and the quality depends on the model's ability to summarize effectively.

# Which Memory Type to Use in which Scenario?
# a. Use ConversationBufferMemory when:
# You want to retain the entire conversation history.
# You have a small number of interactions or a model with a large context window.
# Example1: A tutoring assistant that needs to remember all previous explanations.
# Example2: A RAG-based chatbot that needs to keep all retrieved documents in context.
# Example3: A personal assistant that needs to remember all user preferences and past interactions.

# b. Use ConversationBufferWindowMemory when:
# You only need to remember recent interactions.
# You want to save tokens and reduce costs.
# Example1: A customer support chatbot that only needs to remember the last few messages to resolve an issue.
# Example2: A casual chatbot that doesn't require long-term memory.
# Example3: A simple FAQ bot that only needs to remember the current question and answer.

# c. Use ConversationTokenBufferMemory when:
# You want to manage memory based on token limits rather than message count.
# You have a model with a smaller context window and need to optimize token usage.
# Example1: A chatbot that needs to remember recent interactions but also wants to stay within a 4K token limit.
# Example2: A customer support agent that needs to remember recent messages but also wants to include some important older messages without hitting token limits.
# Example3: A personal assistant that needs to remember recent interactions but also wants to include some important older messages without hitting token limits.

# d. Use ConversationSummaryBufferMemory when:
# You want to retain long-term memory without hitting token limits.
# You want to keep the essence of the conversation without needing every detail.
# Example1: An AI therapist that needs to remember the overall context of a patient's history without needing every single message.
# Example2: A project assistant that needs to remember the overall progress and key decisions without needing every single message.
# Example3: A customer support agent that needs to remember the overall context of a customer's issue without needing every single message.
```

| Memory Type                       | What It Remembers | Advantage          | Ideal Use Case         |
| --------------------------------- | ----------------- | ------------------ | ---------------------- |
| `ConversationBufferMemory`        | Full chat history | Full context       | Teaching or RAG chat   |
| `ConversationBufferWindowMemory`  | Last *k* messages | Keeps chat short   | Quick assistants       |
| `ConversationTokenBufferMemory`   | Last *n* tokens   | Saves model tokens | Cost control           |
| `ConversationSummaryBufferMemory` | Summarized past   | Compact memory     | Long-term conversation |

# **Step 4: Define ConversationBufferWindowMemory with a Window Size of 3**





```python
memory = ConversationBufferWindowMemory(k=3)
memory.save_context({"input": "Hello"}, {"output": "How are you?"}) # shall be lost
memory.save_context({"input": "I'm good, thanks"}, {"output": "Great to hear!"}) # shall be lost
memory.save_context({"input": "I'm good, thanks"}, {"output": "Great to hear!"}) # shall be saved
memory.save_context({"input": "I'm good, thanks"}, {"output": "Great to hear!"}) # shall be saved
memory.save_context({"input": "I'm good, thanks"}, {"output": "Great to hear!"}) # shall be saved
memory.load_memory_variables({})
```

```python
# Loop through and print
for key, value in memory.load_memory_variables({}).items():
    print(f"{key}: {value}")
```

```python
memory = ConversationBufferWindowMemory(k = 1)
conversation = ConversationChain(llm=llm, memory = memory, verbose=True)

conversation.predict(input="Hello, my name is Alex")
conversation.predict(input="What is 2+2?")
conversation.predict(input="What is my name?")
conversation.predict(input="Add 10 to the above output")
print(memory.buffer)
```

# **Step 5: Define ConversationTokenBufferMemory with a Maximum Token Limit of 30**




```python
memory = ConversationTokenBufferMemory(llm=llm, max_token_limit=100)
# Keep only as many messages as fit within 100 tokens.
# If the total conversation exceeds that limit, start dropping the oldest ones.

memory.save_context({"input": "Machine Learning is what?!"}, {"output": "Incredible!"})
memory.save_context({"input": "Neural Networks are what?"}, {"output": "Fascinating!"})
memory.save_context({"input": "AI Assistants are what?"}, {"output": "Impressive!"})
```

```python
# Loop through and print
for key, value in memory.load_memory_variables({}).items():
    print(f"{key}: {value}")
```

```python
# Continue from your previous code
memory.save_context({"input": "What is Deep Learning?"}, {"output": "It's a subset of ML using neural networks."})
memory.save_context({"input": "What is Reinforcement Learning?"}, {"output": "It's about learning through rewards and punishments."})
memory.save_context({"input": "What is NLP?"}, {"output": "It enables computers to understand human language."})
memory.save_context({"input": "What is Computer Vision?"}, {"output": "It allows machines to interpret visual data."})
memory.save_context({"input": "What is Generative AI?"}, {"output": "It creates new content like text, images, or music."})
```

```python
# Loop through and print
for key, value in memory.load_memory_variables({}).items():
    print(f"{key}: {value}")
```

# **Step 6: Define ConversationSummaryBufferMemory with a Maximum Token Limit of 100**





```python
schedule = "There is a meeting at 10am with your design team. \
You will need your sketch designs ready. \
11am-2pm have time to work on your LangChain \
project which will go smoothly because Langchain is such a powerful tool. \
At 2pm, lunch at the sushi restaurant with a client who is flying \
in to meet you to understand the latest in AI. \
Be sure to bring your tablet to show the latest LLM demo."

memory = ConversationSummaryBufferMemory(llm=llm, max_token_limit=100)
memory.save_context({"input": "Hi"}, {"output": "Hello"})
memory.save_context({"input": "What's up?"}, {"output": "Not much, just working"})
memory.save_context({"input": "What is on the agenda today?"}, {"output": f"{schedule}"})
```

```python
# Loop through and print
for key, value in memory.load_memory_variables({}).items():
    print(f"{key}: {value}")
```

```python
memory = ConversationSummaryBufferMemory(llm=llm, max_token_limit=40)

conversation = ConversationChain(llm=llm, memory = memory, verbose=True)

conversation.predict(input="Hello, my name is Alex")
conversation.predict(input="What is 2+2?")
conversation.predict(input="What is my name?")
conversation.predict(input="Add 10 to the above output")
print(memory.buffer)
```

```python
# Loop through and print
for key, value in memory.load_memory_variables({}).items():
    print(f"{key}: {value}")
```
